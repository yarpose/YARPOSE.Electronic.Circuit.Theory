```python
import numpy as np
import matplotlib.pyplot as plt

# محور e
e = np.linspace(0, 1.4, 400)

# نقطهٔ اتصال (knee)
e_knee = 0.25

# بخش اول: افت تند نمایی
i_exp = 1.3 * np.exp(-10 * e)

# مقدار جریان در نقطهٔ اتصال
i_knee = 1.3 * np.exp(-10 * e_knee)

# بخش دوم: خطی با شیب ملایم که از همان نقطه عبور کند
slope = -0.3   # شیب خط
i_lin = i_knee + slope * (e - e_knee)

# ترکیب دو بخش به صورت کاملاً پیوسته
i = np.where(e < e_knee, i_exp, np.maximum(i_lin, 0))

# رسم نمودار
plt.figure(figsize=(6, 4))
plt.plot(e, i, 'k', linewidth=2.5)

plt.xlim(0, 1.4)
plt.ylim(0, 1.3)
plt.xlabel('e', fontsize=14)
plt.ylabel('i', fontsize=14, rotation=0, labelpad=15)
plt.title('curve i - e ', fontsize=13)
plt.grid(False)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# مشخصات موج مثلثی
A = 1           # دامنه موج
T = 2           # دوره موج
t = np.linspace(0, T, 1000)  # بردار زمان

# فرمول تولید موج مثلثی استاندارد
tri_wave = 2*A * np.abs((t/T) - np.floor(t/T + 0.5))

# رسم عمودی موج مثلثی: دامنه در محور x و زمان در محور y
plt.plot(tri_wave, t)
plt.title('شکل موج مثلثی بصورت عمودی')
plt.xlabel('دامنه')
plt.ylabel('زمان')
ax1.axvline(x=1, color='red') 
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import sawtooth

# زمان نمونه ها
t = np.linspace(0, 1, 1000)

# ایجاد شکل موج مثلثی با دامنه 1 از -1 تا 1
tri_wave = sawtooth(2 * np.pi * 3 * t, 0.5)  # 5 هرتز، 0.5 = موج مثلثی

# مقیاس کردن دامنه به میلی آمپر از -100mA تا 100mA
tri_wave_mA = tri_wave * 100  # 100mA دامنه

# رسم شکل موج
plt.plot(t, tri_wave_mA,color='black')
plt.title('Triangle Waveform Amplitude ±100mA')
plt.xlabel('Time [s]')
plt.ylabel('Current [mA]')
plt.axhline(0, color='black', linewidth=2)
plt.axvline(0, color='black', linewidth=2)
plt.show()
```
