```python
import numpy as np
import matplotlib.pyplot as plt

# Parameters (example values for NMOS)
Vth = .4         # Threshold voltage (V)
Vds = 2.0         # Drain-source voltage (V)
mu_n = 400e-4     # Mobility (cm^2/Vs)
Cox = 3e-8        # Oxide capacitance per unit area (F/cm^2)
W = 10e-4         # Width (cm)
L = 1e-4          # Length (cm)

beta = mu_n * Cox * (W / L)  # Process transconductance parameter

# VGS sweep range
Vgs = np.linspace(0, 3, 300)

# Initialize Ids array
Ids = np.zeros_like(Vgs)

# Calculate Ids for each Vgs
for i, Vg in enumerate(Vgs):
    if Vg < Vth:
        Ids[i] = 0  # Cutoff region
    elif Vds < (Vg - Vth):
        # Linear (triode) region
        Ids[i] = beta * ((Vg - Vth) * Vds - 0.5 * Vds**2)
    else:
        # Saturation region
        Ids[i] = 0.5 * beta * (Vg - Vth)**2

plt.plot(Vgs, Ids,color='black',linestyle='--')
# Parameters (example values for NMOS)
Vth1 = 4         # Threshold voltage (V)
Vds1 = 2         # Drain-source voltage (V)
mu_n = 400e-4     # Mobility (cm^2/Vs)
Cox = 3e-8        # Oxide capacitance per unit area (F/cm^2)
W = 10e-4         # Width (cm)
L = 1e-4          # Length (cm)

beta = mu_n * Cox * (W / L)  # Process transconductance parameter

# VGS sweep range
Vgs1 = np.linspace(0, 6, 100)

# Initialize Ids array
Ids1 = np.zeros_like(Vgs1)

# Calculate Ids for each Vgs
for i, Vg1 in enumerate(Vgs1):
    if Vg1 < Vth1:
        Ids1[i] = 0  # Cutoff region
    elif Vds1 < (Vg1 - Vth1):
        # Linear (triode) region
        Ids1[i] = beta * ((Vg1 - Vth1) * Vds1 - 0.5 * Vds1**2)
    else:
        # Saturation region
        Ids1[i] = 0.5 * beta * (Vg1 - Vth1)**2

plt.plot(Vgs1, Ids1,color='black',linestyle='--')
plt.plot([3.5, 3], [0, 3e-8], color='black', linestyle='-', label='Membership = 2')
plt.plot([0, 5.3], [1e-8, 1e-8], color='black', linestyle='-', label='Membership = 2')
plt.plot([0, 3.25], [1.5e-8, 1.5e-8], color='black', linestyle='-', label='Membership = 2')
plt.plot([0, 2.2], [2e-8, 2e-8], color='black', linestyle='-', label='Membership = 2')
plt.text(3.2,-.2e-8, ' $ E{bb} $ ', fontsize=10)
plt.text(3,3.2e-8, ' $R{bb} $ ', fontsize=10)
plt.text(3,3.9e-8, ' $ e{c} $ ', fontsize=10)
plt.text(6,2.5e-8, ' $ e{c1} $ ', fontsize=10)
plt.axhline(0, color='black', linewidth=1)
plt.axvline(0, color='black', linewidth=1)
plt.xlabel('$V_{GS}$ (V)')
plt.ylabel('$I_{DS}$ (A)')
plt.title('MOSFET $I_{DS}$ vs $V_{GS}$ Characteristic')
plt.show()
```

```python
import matplotlib.pyplot as plt
from matplotlib import rcParams
import PySpice.Logging.Logging as Logging
logger = Logging.setup_logging()


from PySpice.Probe.Plot import plot
from PySpice.Spice.Netlist import Circuit
from PySpice.Unit import *

rcParams['font.weight'] = 'bold'
figure, ax1 = plt.subplots(figsize=(15, 5))

circuit = Circuit('v-i curve')
circuit.model('BAV21','D' ,IS = 100@u_uA , N = 1.906  ,BV = 110@u_V, IBV = 0.0001@u_V,  VJ = 20@u_V ,CJO = 7.048E-13 ,RS = 20@u_Ohm ,TT = 3.48E-9)
Vi = circuit.V('i', '1', circuit.gnd, 0@u_V)
circuit.R(1, '1', '2', 10@u_Ohm)
circuit.D(1, '2',circuit.gnd , model = 'BAV21')
circuit.I(1, circuit.gnd,'2', 300@u_uA)
simulator = circuit.simulator(temperature=25, nominal_temperature=25)
analysis = simulator.dc(Vi=slice(-2,8, 0.1))
ax1.plot(analysis['1'], u_mA(-analysis.Vi),'black',linewidth=2,linestyle='-')

ax1.set_xlim(-2, 4)
ax1.set_ylim(-1, 2)
ax1.set_xlabel('V [V]')
ax1.set_ylabel('I [mA]')
ax1.text(0.3, 1.5, 'r', ha='center', va='center', fontsize=12)
ax1.text(.1, -0.2, '$-Io$', ha='center', va='center', fontsize=12)
plt.axhline(0, color='black', linewidth=1 )
plt.axvline(0, color='black', linewidth=1)
plt.tight_layout()
plt.show()
```
