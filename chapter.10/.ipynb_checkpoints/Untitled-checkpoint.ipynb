```python
import numpy as np
import matplotlib.pyplot as plt

# Set up the figure and axes
fig, ax = plt.subplots(figsize=(7, 7))
ax.set_aspect('equal')
ax.set_xlim(-2, 2)
ax.set_ylim(-2, 2)

# Draw axes
ax.axhline(0, color='black', linewidth=1)
ax.axvline(0, color='black', linewidth=1)
ax.set_xlabel('$e_x$')
ax.set_ylabel('$i$')

# Draw family of spirals for various k
theta = np.linspace(-np.pi, np.pi, 400)
k_values = [0, 1, -1, -0.5, -2, -3]
for k in k_values:
    r = np.tan(theta/2)**k
    x = r * np.cos(theta)
    y = r * np.sin(theta)
    # Avoid plotting where r is infinite or NaN
    mask = np.isfinite(x) & np.isfinite(y) & (np.abs(x) < 10) & (np.abs(y) < 10)
    ax.plot(x[mask], y[mask], label=f'$k={k}$')

# Draw k = infinity (the straight line)
ax.plot(np.zeros_like(theta), np.tan(theta/2), 'k--', label='$k=\\infty$')

# Add labels for some curves
ax.text(1.5, 1.2, '$k=0$', fontsize=12)
ax.text(1.2, 1.7, '$k=1$', fontsize=12)
ax.text(-1.7, 1.2, '$k=-1$', fontsize=12)
ax.text(-1.7, -1.2, '$k=-2$', fontsize=12)
ax.text(1.2, -1.7, '$k=-3$', fontsize=12)
ax.text(0.1, 1.8, '$i$', fontsize=14)
ax.text(1.8, 0.1, '$e_x$', fontsize=14)

# Hide default ticks
ax.set_xticks([])
ax.set_yticks([])

# Add legend (optional)
# ax.legend()

plt.title(r'Family of Spirals: $R=-1$')
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Ellipse parameters
a = 3.5  # semi-major axis
b = 2    # semi-minor axis

# Ellipse points
t = np.linspace(0, 2*np.pi, 300)
x = a * np.cos(t)
y = b * np.sin(t)

fig, ax = plt.subplots(figsize=(6, 4))
ax.plot(x, y, color='black')  # Ellipse

# Draw direction arrows along the ellipse
arrow_indices = [40, 120, 200]
for idx in arrow_indices:
    dx = x[idx+1] - x[idx]
    dy = y[idx+1] - y[idx]
    ax.arrow(x[idx], y[idx], dx, dy, shape='full', lw=0, length_includes_head=True,
             head_width=0.15, head_length=0.25, color='black')

# Draw axes
ax.axhline(0, color='black', lw=1)
ax.axvline(0, color='black', lw=1)

# Add dashed vertical line at x = a/2 (for E)
ax.axvline(a/2, color='black', linestyle='dashed', lw=1)

# Add labels
ax.text(0, -0.3, 'O', fontsize=14, ha='center', va='top')
ax.text(a/2, -0.3, 'E', fontsize=14, ha='center', va='top')
ax.text(a+0.2, 0, '$e_x$', fontsize=14, ha='left', va='center')
ax.text(0, b+0.3, 'i', fontsize=14, ha='center', va='bottom')

# Set aspect ratio and limits
ax.set_aspect('equal')
ax.set_xlim(-a-0.5, a+0.5)
ax.set_ylim(-b-0.5, b+0.5)
ax.axis('off')  # Hide axes

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Parameters
L = 1
C = 1
I = 1
E = 1
omega_0 = 1 / np.sqrt(L * C)

# Time axis
t = np.linspace(0, 2 * np.pi * np.sqrt(L * C), 500)
y = E - I * np.sqrt(L / C) * np.sin(omega_0 * t)

# Plot the sinusoidal curve
plt.figure(figsize=(6, 8))
plt.plot(t, y, color='black')

# Draw axes
plt.axhline(E, color='black', linewidth=1)
plt.axvline(0, color='black', linewidth=1)

# Dashed vertical lines at t = pi*sqrt(LC) and t = 2pi*sqrt(LC)
for n in [1, 2]:
    x = n * np.pi * np.sqrt(L * C)
    plt.axvline(x, color='black', linestyle='dashed', linewidth=1)
    plt.text(x + 0.1, E - 1.2, f"${n}\\pi\\sqrt{{LC}}$", fontsize=14, va='center')

# Dashed horizontal lines at y = E - I*sqrt(L/C) and y = E + I*sqrt(L/C)
for sign, y_offset in [(-1, E - I * np.sqrt(L / C)), (1, E + I * np.sqrt(L / C))]:
    plt.axhline(y_offset, color='black', linestyle='dashed', linewidth=1)
    plt.text(0.1, y_offset, f"$E{sign:+}I\\sqrt{{L/C}}$", fontsize=14, va='center')

# Labels
plt.text(-0.5, E, "O", fontsize=14, ha='right', va='center')
plt.text(0, E + 0.2, "E", fontsize=14, ha='center', va='bottom')
plt.text(t[-1] + 0.2, E, "$e_x$", fontsize=14, ha='left', va='center')
plt.text(-0.5, y.min() - 0.2, "$t'$", fontsize=14, ha='right', va='center')

# Annotate the equation
plt.annotate(r'$(E - e_x) = I\sqrt{\frac{L}{C}}\sin \omega_0 t$', 
             xy=(t[100], y[100]), xytext=(t[100]+1, y[100]-0.5),
             arrowprops=dict(arrowstyle="->", color='black'),
             fontsize=14)

# Remove axes
plt.axis('off')
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Parameters
L = 1
C = 1
I = 1
omega_0 = 1 / np.sqrt(L * C)

# Time axis
t = np.linspace(0, 2 * np.pi * np.sqrt(L * C), 500)
i = I * np.cos(omega_0 * t)

# Plot the cosine curve
plt.figure(figsize=(6, 6))
plt.plot(t, i, color='black')

# Draw axes
plt.axhline(0, color='black', linewidth=1)
plt.axvline(0, color='black', linewidth=1)

# Dashed vertical lines at t = pi*sqrt(LC) and t = 2pi*sqrt(LC)
for n in [1, 2]:
    x = n * np.pi * np.sqrt(L * C)
    plt.axvline(x, color='black', linestyle='dashed', linewidth=1)
    plt.text(x + 0.1, -1.2, f"${n}\\pi\\sqrt{{LC}}$", fontsize=14, va='center')

# Dashed horizontal lines at y = I and y = -I
for sign in [1, -1]:
    plt.axhline(sign * I, color='black', linestyle='dashed', linewidth=1)
    plt.text(0.1, sign * I, f"${sign:+}I$", fontsize=14, va='center')

# Labels
plt.text(-0.3, 0, "O", fontsize=14, ha='right', va='center')
plt.text(t[-1] + 0.2, 0, "$t'$", fontsize=14, ha='left', va='center')
plt.text(0, 1.1 * I, "$i$", fontsize=14, ha='center', va='bottom')

# Annotate the equation
plt.annotate(r'$i = I \cos \omega_0 t$', 
             xy=(t[60], i[60]), xytext=(t[60]+1, i[60]+0.5),
             arrowprops=dict(arrowstyle="->", color='black'),
             fontsize=14)

# Remove axes
plt.axis('off')
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Define the data for the plot
# Define the circles
circles = [
    {'radius': 0.5, 'linestyle': 'solid'},
    {'radius': 1, 'linestyle': 'solid'},
    {'radius': 1.5, 'linestyle': 'solid'},
    {'radius': 2, 'linestyle': 'solid'},
]

# Define the lines
lines = [
    {'start': (0, 0), 'end': (0, 3), 'label': 'k=-1/2', 'position': (0.5, 3)},
    {'start': (0, 0), 'end': (0, -3), 'label': '', 'position': (0, 0)},
    {'start': (0, 0), 'end': (3, 0), 'label': 'k=∞', 'position': (3, 0.2)},
    {'start': (0, 0), 'end': (-3, 0), 'label': '', 'position': (0, 0)},
    {'start': (0, 0), 'end': (2.8, -1.2), 'label': 'k=0', 'position': (3.2, -1.3)},
    {'start': (0, 0), 'end': (-2.8, 1.2), 'label': 'v=1-i', 'position': (-3.5, 1.2)},
]

# Define the points
points = [
    {'position': (0, 1), 'text': '1'},
    {'position': (0, -1), 'text': '-1'},
]
# Define the isocline lines data
isoclines = [
    {'start': (0, 0), 'end': (-0.7, 3), 'label': 'k=-1', 'position': (-1, 3)},
    {'start': (0, 0), 'end': (-1.4, 3), 'label': 'k=-2', 'position': (-1.7, 3)},
    {'start': (0, 0), 'end': (-2.1, 3), 'label': 'k=-3', 'position': (-2.4, 3)},
    {'start': (0, 0), 'end': (0.7, -3), 'label': 'k=1', 'position': (1, -3)}
]
# Plot the data
fig, ax = plt.subplots(figsize=(6, 6))

# Plot circles
for circle in circles:
    c = plt.Circle((0, 0), circle['radius'], linestyle=circle['linestyle'], edgecolor='black', fill=False)
    ax.add_artist(c)

# Plot lines
for line in lines:
    ax.plot([line['start'][0], line['end'][0]], [line['start'][1], line['end'][1]], color='black', linestyle='-', linewidth=1)
    x = line['end'][0]
    y = line['end'][1]
    ax.text(line['position'][0], line['position'][1], line['label'], fontsize=8)

# Plot isocline lines
for line in isoclines:
    x = np.linspace(0, line['end'][0], 10)
    y = np.linspace(0, line['end'][1], 10)
    for i in range(len(x) - 1):
        mid_x = (x[i] + x[i+1]) / 2
        mid_y = (y[i] + y[i+1]) / 2
        dx = (x[i+1] - x[i]) * 0.4
        dy = (y[i+1] - y[i]) * 0.4
        ax.plot([mid_x - dx, mid_x + dx], [mid_y - dy, mid_y + dy], color='black', linewidth=1)
        ax.text(line['position'][0], line['position'][1], line['label'], fontsize=8)
# Plot points and text
for point in points:
    ax.plot(point['position'][0], point['position'][1], marker='o', color='black', markersize=3)
    ax.text(point['position'][0] + 0.1, point['position'][1] + 0.1, point['text'], fontsize=8)

# Set the axis limits
ax.set_xlim(-4, 4)
ax.set_ylim(-4, 4)

# Add the axis labels
ax.set_xlabel('ex')
ax.set_ylabel('i')

# Add grid lines
ax.grid(True)
ax.axhline(0, color='black', linewidth=0.5)
ax.axvline(0, color='black', linewidth=0.5)
# Remove ticks
ax.set_xticks([])
ax.set_yticks([])
# Add title
ax.set_title('Phase-plane trajectories by the method of isoclines')
# Show the plot
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp

# Circuit parameters
L = 1.0  # Inductor (henry)
R = 4.0  # Resistor (ohms)
C = 1.0  # Capacitor (farad)
V_source = 1.0  # Source voltage (volts)
V_load = 1.0    # Load/power supply voltage (volts)

# Define the system of differential equations
def circuit_ode(t, y):
    """
    y[0]: Current through the circuit (A)
    y[1]: Voltage across the capacitor (V)
    """
    i, v_c = y
    
    # Differential equations from Kirchhoff's laws
    # V_source - L*di/dt - R*i - v_c - V_load = 0
    di_dt = (V_source - R*i - v_c - V_load) / L
    
    # Capacitor equation: dv_c/dt = i/C
    dv_c_dt = i / C
    
    return [di_dt, dv_c_dt]

# Time span for simulation
t_span = (0, 15)  # Simulate from 0 to 15 seconds
t_eval = np.linspace(0, 15, 1500)  # Points at which to evaluate the solution

# Initial conditions: i(0) = 0, v_c(0) = 0
initial_conditions = [0.0, 0.0]

# Solve the system of ODEs
solution = solve_ivp(circuit_ode, t_span, initial_conditions, 
                    t_eval=t_eval, method='RK45')

# Extract time and solution
t = solution.t
current = solution.y[0]
v_capacitor = solution.y[1]

# Calculate theoretical steady-state values
i_steady = (V_source - V_load) / R  # When capacitor is fully charged
v_c_steady = V_source - V_load - R*i_steady

# Create visualization of the circuit behavior
plt.figure(figsize=(12, 10))

# Plot the current
plt.subplot(3, 1, 1)
plt.plot(t, current)
plt.axhline(y=i_steady, color='r', linestyle='--', label=f'Steady State: {i_steady:.3f} A')
plt.title('Current Flowing Through the Circuit (Into 1V Power Supply)')
plt.xlabel('Time (s)')
plt.ylabel('Current (A)')
plt.grid(True)
plt.legend()

# Plot the capacitor voltage
plt.subplot(3, 1, 2)
plt.plot(t, v_capacitor)
plt.axhline(y=v_c_steady, color='r', linestyle='--', label=f'Steady State: {v_c_steady:.3f} V')
plt.title('Voltage Across Capacitor')
plt.xlabel('Time (s)')
plt.ylabel('Voltage (V)')
plt.grid(True)
plt.legend()

# Plot the power delivered to the 1V power supply
power_to_load = V_load * current
plt.subplot(3, 1, 3)
plt.plot(t, power_to_load)
plt.axhline(y=V_load*i_steady, color='r', linestyle='--', label=f'Steady State: {V_load*i_steady:.3f} W')
plt.title('Power Delivered to 1V Power Supply')
plt.xlabel('Time (s)')
plt.ylabel('Power (W)')
plt.grid(True)
plt.legend()

plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
from PySpice.Spice.Netlist import Circuit
from PySpice.Unit import *

# Define the circuit
circuit = Circuit('Series RLC Step Response')

# Voltage source: Step from 0V to 1V at t=0
circuit.PulseVoltageSource('input', 'in', circuit.gnd, 
                           initial_value=0@u_V, pulsed_value=1@u_V, 
                           pulse_width=10@u_s, period=20@u_s)

# Inductor: 1 H
circuit.L('1', 'in', 'n1', 1@u_H)

# Resistor: 2 Ω
circuit.R('1', 'n1', 'n2', 2@u_Ohm)

# Capacitor: 1 F
circuit.C('1', 'n2', circuit.gnd, 1@u_F)

# Simulation: Transient analysis for 5 seconds
from PySpice.Probe.Plot import plot
from PySpice.Spice.Library import SpiceLibrary
from PySpice.Spice.Simulation import *

simulator = circuit.simulator(temperature=25, nominal_temperature=25)
analysis = simulator.transient(step_time=1e-3, end_time=5)
figure,ax = plt.subplots(figsize=(10, 5))
plt.plot(analysis.time, u_A(-analysis.L1),color='black')
plt.axhline(0, color='black', linewidth=1)
plt.axvline(0, color='black', linewidth=1)
plt.title('Voltage Across Capacitor')
plt.xlabel('Time [s]')
plt.ylabel('Capacitor Voltage [V]')
ax.plot([0,5], [-.37,-.37], color='black', linestyle='--', label='Membership = 1')
ax.text(3 ,-.35, r'$ \frac {1}{e}=0.36 $', ha='center', va='center')
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Given parameters
V_source = 1  # volts
R = 4         # ohms
L = 1         # henry
C = 1         # farad

# Time array
t = np.linspace(0, 7, 500)

# Calculate damping factor and natural frequency
alpha = R / (2*L)
omega_0 = 1 / np.sqrt(L*C)

# Damped natural frequency
omega_d = np.sqrt(omega_0**2 - alpha**2) if omega_0 > alpha else 0

# Voltage across capacitor as a function of time (underdamped case)
if omega_d > 0:
    Vc = V_source * (1 - np.exp(-alpha * t) * (np.cos(omega_d * t) + (alpha / omega_d) * np.sin(omega_d * t)))
else:
    # Overdamped or critically damped case
    Vc = V_source * (1 - np.exp(-alpha * t) * (1 + alpha * t))

# Plotting with rotated time axis 90 degrees clockwise
plt.figure(figsize=(6,6))
plt.plot(Vc, t)  # Key part for rotation: swap x and y values
plt.xlabel('Voltage across Capacitor (V)')
plt.ylabel('Time (s)')
plt.title('Voltage across Capacitor vs Time (Rotated Axis)')
plt.gca().invert_xaxis()  # Invert x-axis to complete the rotation effect
plt.grid(True)
plt.show()
```
