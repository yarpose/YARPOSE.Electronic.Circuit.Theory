$3.49$. A gas diode is placed in series with a 100 kilohm resistance and the combination is connected to a variable voltage source. The voltage is slowly raised from zero. The circuit draws negligible current until the source voltage reaches 100 volts, at which time the current jumps to nearly 1 ma.
