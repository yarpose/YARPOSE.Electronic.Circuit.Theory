```python
import numpy as np
import matplotlib.pyplot as plt

# Circuit parameters
R = 1000  # Resistance in ohms
C = 1e-6  # Capacitance in farads

# Frequency range
frequencies = np.logspace(1, 6, 400)  # 10 Hz to 1 MHz

# Transfer function for a high-pass RC filter: s / (s + 1) where s = jwRC
def high_pass_transfer_function(f, R, C):
    w = 2 * np.pi * f
    s = 1j * w * R * C
    return s / (1 + s)

# Calculate the transfer function
transfer_function = high_pass_transfer_function(frequencies, R, C)

# Calculate magnitude and phase (in degrees)
magnitude = np.abs(transfer_function)
phase = np.angle(transfer_function)

# Plotting
figure, ax1 = plt.subplots(figsize=(10, 8))

ax1.semilogx(frequencies, 20 * np.log10(magnitude), color='black')  # Magnitude in dB
ax1.set_ylabel('|E2/E1|')
ax1.set_title('w')
ax1.plot([0, 100], [-5, -5], color='black', linestyle='--', label='Membership = 2')
ax1.plot([100, 100], [-24, -5], color='black', linestyle='--', label='Membership = 2')
ax1.plot([9, 10000], [0, 0], color='black', linestyle='--', label='Membership = 2')
ax1.text( 100, -25, '$W_c$', ha='center', va='center')
ax1.text( 10,-4, '$0.707K$', ha='center', va='center')
ax1.text( 1000,-24, '$√(ωc.ωs)$', ha='center', va='center')
ax1.text( 7, 0, '$K$', ha='center', va='center')
ax1.text( 1000,-15, r'$ ωc= \frac {1}{τc}= \frac {1}{C[Rg+rp||Rb]}$', ha='center', va='center')
 
```

```python
from scipy.signal import butter, lfilter, buttord, freqz
import numpy as np
import matplotlib.pyplot as plt

# Generate sample data
t = np.arange(0, 8192, 1)
data = 1 + np.sin(.15 * np.pi * t + np.pi/2)

# Design the bandpass filter
b, a = butter(6, [0.15, 0.85], btype='band')

# Apply the filter to the data
y = lfilter(b, a, data)

# Take the FFT of the filtered data
y = np.fft.rfft(y)
y = np.abs(y)

# Take the FFT of the original data
x = np.fft.rfft(data)
x = np.abs(x)

# Calculate the frequency response
w, h = freqz(b, a, worN=4096)
plt.figure(4)
plt.plot(np.arange(0, 4096, 1), np.abs(h), color='black')
plt.title('Frequency Response of the Filter')
plt.text( 0,0.85, '$0.707K$', ha='center', va='center')
plt.plot([0, 4000], [0.8, 0.8], color='black', linestyle='--', label='Membership = 2')
plt.plot([620, 620], [0, 0.8], color='black', linestyle='--', label='Membership = 2')
plt.plot([3450, 3450], [0, 0.8], color='black', linestyle='--', label='Membership = 2')
plt.text( 620,-.1, '$w_c$', ha='center', va='center')
plt.text( 3450,-.1, '$W_s$', ha='center', va='center')
plt.show()
```
