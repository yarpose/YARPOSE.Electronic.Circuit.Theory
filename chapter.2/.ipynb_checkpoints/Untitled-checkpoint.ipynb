```python
import matplotlib.pyplot as plt
from matplotlib import rcParams
import PySpice.Logging.Logging as Logging
logger = Logging.setup_logging()

from PySpice.Probe.Plot import plot
from PySpice.Spice.Netlist import Circuit
from PySpice.Unit import *
figure, ax1 = plt.subplots(figsize=(15, 5))

circuit = Circuit('Transitional characteristic1')
circuit.model('BAV21','D' ,IS = .1@u_mA   ,BV = 293.10@u_V, IBV = 0.001293@u_A , RS = 1@u_Ohm )
Vi = circuit.V('i', '1', circuit.gnd, 0@u_V)
circuit.R(1, '1', '2', 1@u_kOhm)
circuit.R(2, '3', '4', 10@u_Ohm)
circuit.D(1, '2','3', model = 'BAV21')
circuit.V(1, '4', circuit.gnd, 2@u_V)
circuit.D(2, '5', '2', model = 'BAV21')
circuit.V(2, circuit.gnd, '6', 2@u_V)
circuit.R(3, '5', '6', 10@u_Ohm)
simulator = circuit.simulator(temperature=25, nominal_temperature=25)
analysis = simulator.dc(Vi=slice(-5, 5, .1))
ax1.plot(analysis['1'], analysis['2'],'black',linewidth=5)
ax1.axhline(0, color='black', linewidth=1)
ax1.axvline(0, color='black', linewidth=1)
circuit = Circuit('Transitional characteristic1')
circuit.model('BAV21','D' ,IS = 1@u_mA   ,BV = 293.10@u_V, IBV = 0.001293@u_A , RS = 100@u_Ohm )
Vi = circuit.V('i', '1', circuit.gnd, 0@u_V)
circuit.R(1, '1', '2', 8@u_kOhm)
circuit.R(2, '3', '4', 1@u_kOhm)
circuit.D(1, '2','3', model = 'BAV21')
circuit.V(1, '4', circuit.gnd, 4@u_V)
circuit.D(2, '5', '2', model = 'BAV21')
circuit.V(2, circuit.gnd, '6', 4@u_V)
circuit.R(3, '5', '6', 1@u_Ohm)
simulator = circuit.simulator(temperature=25, nominal_temperature=25)
analysis = simulator.dc(Vi=slice(-5, 5, .1))
ax1.plot(analysis['1'], analysis['2'],'black',linewidth=5,linestyle='--')
ax1.axhline(0, color='black', linewidth=1)
ax1.axvline(0, color='black', linewidth=1)
circuit = Circuit('Transitional characteristic1')
circuit.model('BAV21','D' ,IS = 1@u_mA   ,BV = 293.10@u_V, IBV = 0.001293@u_A , RS = 100@u_Ohm )
Vi = circuit.V('i', '1', circuit.gnd, 0@u_V)
circuit.R(1, '1', '2', 8@u_kOhm)
circuit.R(2, '3', '4', 1@u_Ohm)
circuit.D(1, '2','3', model = 'BAV21')
circuit.V(1, '4', circuit.gnd, 1.5@u_V)
circuit.D(2, '5', '2', model = 'BAV21')
circuit.V(2, circuit.gnd, '6', 1.5@u_V)
circuit.R(3, '5', '6', 1@u_Ohm)
simulator = circuit.simulator(temperature=25, nominal_temperature=25)
analysis = simulator.dc(Vi=slice(-5, 5, .1))
ax1.plot(analysis['1'], analysis['2'],'black',linewidth=5,linestyle='--')
ax1.axhline(0, color='black', linewidth=1)
ax1.axvline(0, color='black', linewidth=1)
ax1.text(4, 1, 'forward bias', ha='center', va='center')
ax1.text(4, 2.5, 'Zero bias', ha='center', va='center')
ax1.text(4, 3.2, 'Reverse bias', ha='center', va='center')
```

```python
import matplotlib.pyplot as plt
from matplotlib import rcParams
import matplotlib.image as mpimg
import numpy as np

import PySpice.Logging.Logging as Logging
logger = Logging.setup_logging()


from PySpice.Probe.Plot import plot
from PySpice.Spice.Netlist import Circuit
from PySpice.Unit import *

rcParams['font.weight'] = 'bold'

figure, ax1 = plt.subplots(figsize=(15, 5))

circuit = Circuit('v-i curve')
circuit.model('BAV21','D' ,IS = 100@u_uA , N = 1.906  ,BV = 110@u_V, IBV = 0.0001@u_V,  VJ = 20@u_V ,CJO = 7.048E-13 ,RS = 20@u_Ohm ,TT = 3.48E-9)
Vi = circuit.V('i', '1', circuit.gnd, 0@u_V)
circuit.R(1, '2', '3', 1@u_Ohm)
circuit.D(1,'2' ,'1', model = 'BAV21')
circuit.V(1, circuit.gnd,'3', 1@u_V)
circuit.R(2, '1', '4', 1@u_Ohm)
circuit.D(2,'4',circuit.gnd , model = 'BAV21')
circuit.I(1, circuit.gnd,'4', 200@u_uA)
circuit.V(2,'6', circuit.gnd, 2@u_V)
circuit.R(3, '5', '6', 1@u_Ohm)
circuit.D(3,'1','5' , model = 'BAV21')
simulator = circuit.simulator(temperature=25, nominal_temperature=25)
analysis = simulator.dc(Vi=slice(-2,8, 0.1))
ax1.plot(analysis['1'], u_mA(-analysis.Vi),'black',linewidth=2,linestyle='-')


ax1.set_xlim(-4, 4)
ax1.set_ylim(-4, 4)
ax1.set_xlabel('V [V]')
ax1.set_ylabel('I [mA]')
ax1.text(.1, -0.2, '$-Io$', ha='center', va='center', fontsize=12)
ax1.text(-1, 0.2, '$-Vz$', ha='center', va='center', fontsize=12)
ax1.axhline(0, color='black', linewidth=1 )
ax1.axvline(0, color='black', linewidth=1)
```
