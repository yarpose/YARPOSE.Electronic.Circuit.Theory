All digital designs are based on primitive logic operations. The first task in
our study of VLSI will be to create electronic logic gates that can be used
as building blocks in complex switching networks.

an open switch. This means that there is no relationship between the two

sides. The opposite case is a closed switch where we visualize the top portion
of the switch being “pushed down” as shown in Figure 2.1(b). This

### Figure 2.24 Operation o f the CMOS NOT 

• Use a complementary nFET/pFET pair for each input
• Connect the output node to the power supply VD D through pFETs
• Connect the output node to ground through nFETs, and
• Insure that the output is always a well-defined high or low voltage
This set o f guidelines helps us design logic circuits that have input and
output characteristics which are compatible with the N O T gate.
The logic symbol and truth table for the NOR2 gate are provided in Fig

$$g(x.y)=(x+y) ̅ $$ $$(2.36)$$

= (1,1) yields an output o f g = 1.

The procedure for designing the transistor circuitry for a CM O S logic
gate can be summarized by the following steps.
                                                    
• Construct the logic diagram using basic AOI or OAI structuring.
Deeper nesting, such a OAOI and AOAI, is allowed.

• Use the gate-nFET relations summarized in Figure 2.43 to construct
the nFET logic circuit between the output and ground.

• To obtain the topology of the pFET array, start with the original logic
diagram and push the bubble back toward the inputs using the
DeMorgan rules. Continue the backward pushing until ev eiy input
is bubbled. The pFET circuitry between the output and VD D is then
obtained using the rules in Figure 2.51.

Note that both the nFETs and the pFETs are wired such that parallel-con
nected transistors give the OR operation, w h ile series-con n ected FETs
provide the AND operation. The only difference betw een the tw o is that
nFETs are assert-high devices while pFETs are assert-low (bubbled-input)
switches.

```python
import schemdraw
import schemdraw.elements as elm
import matplotlib.pyplot as plt
from schemdraw import dsp
from schemdraw import logic
with schemdraw.Drawing(unit=2) as d:
    Nand= d.add(logic.Nand().right().label('a \n b', loc='left', color='black').label(r'($\bar {a}$.$\bar {b}$)','right'))
    d += (l1 :=elm.Arrow(double=True).right(d.unit*1).at((6,0)).label('$aquvalent$','top').label('$(a) Nand-or$','bottom'))
    Or= d.add(logic.Or().right().at((10,0)))
    d += (l1 :=elm.Line().right(d.unit*.4).at((9.2,-0.25)).label('$a$','left').idot().dot(open=True))
    d += (l2 :=elm.Line().right(d.unit*.4).at((9.2,0.25)).label('$b$','left').idot().dot(open=True))
    d += (l1 :=elm.Line().right(d.unit*.4).at((11.9,0)).label(r'($\bar {a}$.$\bar {b}$)','right').dot())
    Nor= d.add(logic.Nor().right().label('a \n b', loc='left', color='black').label(r'($\bar {a}$+$\bar {b}$)','right').at((0,-4)))
    d += (l1 :=elm.Arrow(double=True).right(d.unit*1).at((6,-4)).label('$aquvalent$','top').label('$(b) Nor-And$','bottom'))
    And= d.add(logic.And().right().at((10,-4)))
    d += (l1 :=elm.Line().right(d.unit*.4).at((9.2,-3.75)).label('$a$','left').idot().dot(open=True))
    d += (l2 :=elm.Line().right(d.unit*.4).at((9.2,-4.25)).label('$b$','left').idot().dot(open=True))
    d += (l1 :=elm.Line().right(d.unit*.4).at((11.9,-4)).label(r'($\bar {a}$.$\bar {b}$)','right').dot())
```

## Figure 2.58 General naming convention

Generalized complex logic gates provide a u niform basis for creating
different logic operations using a generic gate. As a simple example, con
sider the AOI22 gate shown in Figure 2.59(a). This provides an output of

$AOI22(a, b, c, d ) = (a • b + c • d)^-$ $(2.75)$ 

To create an XOR circuit, we can define the inputs as show n in Figure
2.59(b), which allows us to write 

a ⊕ b = AO I22(a, b, $\bar {a}$, $\bar {b}$) $(2.76)$ 

Using the same reasoning, the XNOR function can be obtained using

# 2.6 Clocking and Dataflow Control

Synchronous digital design relies on the ability to control the flow of date
usin g a clockin g signal ∅. Th e sw itching characteristics of TGs can
used to provide a simple approach to system clocking. Since complemen
tary signals are required to switch a TG, both ∅and $\bar{∅}$ are used in this try
of design; waveform s are shown in Figure 2.65. The period T is the tims
seconds needed for one complete cycle. The frequency f is defined by

$f=\frac {1}{T}$ $(2.84)$

and has units o f Hertz [Hz] = [1/sec], where 1 hertz means one cycle
completed in 1 second. We will assume that the clock is at a logic 1vi
o f on e-h alf o f the period, and at a logic 0 value for the remaining hai
the period.
Let us exam ine the effect o f applying the complementary clocks
transm ission gate. Figure 2.66(a) shows that when a value o f ∅ = 1
applied to the nFET and $\bar{∅}$ = 0 to the pFET, the TG is On and acts little
closed switch. Reversing these values as in Figure 2.66(b) gives an
switch. Under static conditions, the value of y would not be known white
the switch is opened. However, the electrical characteristics of CMOS
allow us to temporarily hold the value o f y = x for a very short time 
typically, th old is less than 1 second. If we use a high-frequency clock time
the periodic open-closed change occurs at every half clock cycle. Then
can hold the previous value so long as (T/2) < t hold. This provides
accurate time base for controlling data flow in a complex network.

```python
import matplotlib.pyplot as plt
import numpy as np
from matplotlib import rcParams
import PySpice.Logging.Logging as Logging
from PySpice.Probe.Plot import plot
from PySpice.Spice.Library import SpiceLibrary
from PySpice.Spice.Netlist import Circuit
from PySpice.Unit import *

logger = Logging.setup_logging()
circuit = Circuit('Pulse Sources ')

circuit.model('2N7000', 'NMOS', L=0.18E-6, W=.5E-6,LAMBDA = 0.1 ,Kp=0.007, Vto=.3)
circuit.model('2N7100', 'PMOS', L=0.18E-6, W=4E-6,LAMBDA = 0.1 ,Kp=0.007, Vto=-.3)
pulse1 = circuit.PulseVoltageSource('V1', 'D', circuit.gnd, initial_value=1.8@u_V, pulsed_value=0@u_V,delay_time=5@u_ns,rise_time=.01@u_ns,fall_time=.01@u_ns, pulse_width=5@u_ns, period=10@u_ns)
pulse2 = circuit.PulseVoltageSource('V2', 'CP', circuit.gnd, initial_value=0@u_V, pulsed_value=1.8@u_V,delay_time=5@u_ns,rise_time=1@u_ps,fall_time=1@u_ps,pulse_width=5@u_ns, period=10@u_ns) 
circuit.V(1, '1', circuit.gnd, 1.8@u_V)
#NOT1
circuit.MOSFET(1, 'D_', 'D', '1','1', model='2N7100')
circuit.MOSFET(2, 'D_', 'D', circuit.gnd,circuit.gnd, model='2N7000')
simulator = circuit.simulator(temperature=25, nominal_temperature=25)
analysis = simulator.transient(step_time=1@u_ns, end_time=40@u_ns)

fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(10, 6), sharex=True)
ax1.plot(analysis.time, analysis['CP'], 'black', linewidth=2, label='CP')
ax1.set_xlabel('Time [s]', color='black', weight='black', fontsize=10)
ax1.set_ylabel('∅', color='black', weight='black', fontsize=10)
ax2.plot(analysis.time, analysis['D'], 'black', linewidth=2, label='D')
ax2.set_xlabel('Time [s]', color='black', weight='black', fontsize=10)
ax2.set_ylabel(r'$\bar{∅}$', color='black', weight='black', fontsize=10)
ax1.axhline(0, color='black', linewidth=2)
ax1.axvline(0, color='black', linewidth=2)
ax2.axhline(0, color='black', linewidth=2)
ax2.axvline(0, color='black', linewidth=2)
```
